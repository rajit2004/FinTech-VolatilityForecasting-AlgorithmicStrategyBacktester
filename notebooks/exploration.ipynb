# Data Exploration

A quick look at the sample data used by the volatility forecaster. We generate the bundled CSV files, load one symbol, and plot the price alongside its realized volatility.

Run the cells below with the project's virtual environment kernel.

```python
import sys
import pandas as pd
from pathlib import Path

# Make sure the app package is importable when running from a notebook.
ROOT = Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from scripts.generate_sample_data import build_sample_dataset, simulate_ohlcv
from app.features.engineering import build_features

symbol = 'AAPL'
build_sample_dataset([symbol])
df = simulate_ohlcv(symbol)
print(df.head())
print(f'\nTotal bars: {len(df)}')
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))
plt.plot(pd.to_datetime(df['date']), df['close'])
plt.title(f'{symbol} closing price')
plt.xlabel('date')
plt.ylabel('price')
plt.show()
```

```python
data = df.copy()
data['date'] = pd.to_datetime(data['date'])
data = data.set_index('date')
features = build_features(data, horizon=5)

plt.figure(figsize=(12, 4))
plt.plot(features.index, features['realized_vol'], label='realized vol')
plt.plot(features.index, features['ewma_vol'], label='ewma vol')
plt.legend()
plt.title(f'{symbol} realized vs EWMA volatility')
plt.xlabel('date')
plt.ylabel('annualized volatility')
plt.show()
```

The volatility line clearly shows clustering: calm stretches followed by violent stretches. That is the main thing the forecasting models try to capture.
